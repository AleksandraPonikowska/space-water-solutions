# Water retention index :3

```python
import pandas as pd
import numpy as np

def process_wri(file_path):

    df = pd.read_csv(file_path)

    dist_norm = df['dist_to_ditch'].clip(upper=1000) / 1000
    
    base_wri = (df["NDVI"] * 0.30 + 
                df["NDMI"] * 0.40 + 
                df["NDWI"].clip(lower=0) * 0.10 +
                dist_norm * 0.2)

    basin_bonus = np.where(df['in_basin'] == 1, 1.15, 1.0)
    
    sealing_penalty = (df['manholes_300m'] * 0.02).clip(upper=0.3)
    
    df["WRI"] = (base_wri * basin_bonus) - sealing_penalty
    df["WRI"] = df["WRI"].clip(0, 1)
    df["WRI"].round(4)
    
    df.to_csv(file_path, index=False)
    print(f"Bajlando! {file_path}")

REGION = "wroclaw_small"
process_wri(f"../data/{REGION}/merged.csv")
process_wri(f"../data/{REGION}/forecast_1m.csv")
```
